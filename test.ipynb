```python
import torch 

from torch import nn

# patch embedding 利用conv2d投影
#B, C, H, W = x.shape
batch_size = 1 
H = 224
W = 224
C = 3
input = torch.randn((batch_size, C, H, W))
print(f"input: {input.shape}")

patch_size = (16, 16)
embedding_dim = 16 * 16 * 3  
proj= nn.Conv2d(in_channels=C, out_channels=embedding_dim, kernel_size=patch_size, stride=patch_size)
patch_embedding = proj(input)
# embedding_dim * (H/16 *(W/16)
print(patch_embedding.shape)
```
